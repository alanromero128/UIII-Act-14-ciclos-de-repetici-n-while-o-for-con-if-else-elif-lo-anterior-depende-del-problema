Calcula el promedio de notas de M alumnos, donde cada alumno tiene N asignaturas. Usa bucles anidados.



```python
# Solicitud de datos iniciales
m = int(input("Ingrese la cantidad de alumnos (M): "))
n = int(input("Ingrese la cantidad de asignaturas por alumno (N): "))

# Contador para el ciclo de alumnos
contador_alumnos = 0

while contador_alumnos < m:
    contador_alumnos = contador_alumnos + 1
    print("\n--- Alumno", contador_alumnos, "---")

    suma_notas = 0
    contador_asignaturas = 0

    # Ciclo para capturar las notas de cada asignatura
    while contador_asignaturas < n:
        contador_asignaturas = contador_asignaturas + 1
        nota = float(input("Ingrese la nota de la asignatura " + str(contador_asignaturas) + ": "))

        # Validacion simple con if-else
        if nota >= 0 and nota <= 10:
            suma_notas = suma_notas + nota
        else:
            print("Nota inválida, se tomará como 0.")
            # suma_notas no aumenta

    # Cálculo del promedio
    promedio = suma_notas / n

    # Evaluación del desempeño
    print("Promedio del alumno " + str(contador_alumnos) + ": " + str(promedio))

    if promedio >= 9:
        print("Resultado: Excelente")
    elif promedio >= 6:
        print("Resultado: Aprobado")
    else:
        print("Resultado: Reprobado")
```
